```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a numpy array x from 0 to 1 in 100 steps
x = np.linspace(0, 1, 100)
```

```python
# Define functions that return sin(x) and cos(x)
def my_sin(x):
    return np.sin(x)

def my_cos(x):
    return np.cos(x)
```

```python
# Create a multipanel plot: 1 row, 2 columns
fig, (ax1, ax2) = plt.subplots(1, 2)

# Left panel: sin(x)
ax1.plot(x, my_sin(x))
ax1.set_title('sin(x)')
ax1.set_xlabel('x')
ax1.set_ylabel('sin(x)')

# Right panel: cos(x)
ax2.plot(x, my_cos(x))
ax2.set_title('cos(x)')
ax2.set_xlabel('x')
ax2.set_ylabel('cos(x)')

plt.tight_layout()
plt.savefig('session8.pdf')
plt.show()
```
