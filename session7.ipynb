```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a numpy array x from 0 to 1 in 100 steps
x = np.linspace(0, 1, 100)
```

```python
# Define a function that returns exp(x)
def my_exp(x):
    return np.exp(x)
```

```python
# Use the function to set y = exp(x), then plot x vs y
y = my_exp(x)

plt.plot(x, y)
plt.xlabel('Time [milliseconds]')
plt.ylabel('Awesomeness')
plt.savefig('session7.pdf')
plt.show()
```
