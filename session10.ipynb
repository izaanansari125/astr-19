```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Pull 1000 random numbers from a Beta distribution (a=2, b=5)
# The Beta distribution lives between 0 and 1 and has a skewed shape
samples = np.random.beta(2, 5, 1000)
```

```python
# Histogram the random numbers into 100 bins and plot
plt.hist(samples, bins=100)
plt.xlabel('Value')
plt.ylabel('Count')
plt.title('Beta Distribution (a=2, b=5)')
plt.savefig('session10.pdf')
plt.show()
```
